# **Categorical Plot**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## **Strip Plot and Swarm Plot**

```python
tips = sns.load_dataset("tips")
tips
```

```python
#basic strip plot
sns.stripplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
#basic swarn plot
sns.swarmplot(data=tips, x="day", y="total_bill")
plt.show()
```

## **Box Plot**

```python
#Horizontal Boxplot of Total Bill
sns.boxplot(x=tips["total_bill"])
plt.show()
```

```python
#Horizontal Boxplot of multiple columns
sns.boxplot(data=tips[["total_bill", "tip", "size"]], orient="h")
plt.show()
```

```python
#box plots wrt day
sns.boxplot(x="day", y="total_bill", data=tips)
plt.show()
```

```python
#boxplot with hue
sns.boxplot(x="day", y="total_bill", hue="smoker", data=tips)
plt.show()
```

```python
#boxplot with hue(without dodging)
sns.boxplot(x="day", y="total_bill", hue="smoker", data=tips, dodge=False)
plt.show()
```

```python
#changing outline thickness
sns.boxplot(x="day", y="tip", data=tips, linewidth=2)
plt.show()
```

```python
#setting single color for enture boxplot
sns.boxplot(x="day", y="tip", data=tips, color="green")
plt.show()
```

## **Bar Plot**

```python
titanic = sns.load_dataset("titanic")
titanic
```

```python
#simple bar plot between class and fare
sns.barplot(x = 'class', y = 'fare', data = titanic)
plt.show()
```

```python
#Creating a bar plot using hue parameter with two variables.(grouped bar chart)
sns.barplot(x = 'class', y = 'fare', hue = 'sex', data = titanic)
plt.show()
```

```python
#horizontal bar plot
sns.barplot(x = 'fare', y = 'class', hue = 'sex', data = titanic)
plt.show()
```

```python
#Control barplot order by passing an explicit order.
sns.barplot(x = 'class', y = 'fare', data = titanic,
            order = ["Third", "Second", "First"])
plt.show()
```

```python
#plot all the bars in single color
sns.barplot(x = 'class', y = 'fare', data = titanic, color = "salmon")
plt.show()
```

The default error bars show 95% confidence intervals.

```python
#barplot without error bar
sns.barplot(x = 'class', y = 'fare', data = titanic, errorbar=None)
plt.show()
```

## **Count Plot**
Similar to bar plot but displays number of observations in each category rather than computing a statistic for a second variable

```python
tips = sns.load_dataset('tips')
tips
```

```python
# count plot on single categorical variable
sns.countplot(x ='sex', data = tips)
plt.show()
```

```python
# count plot on two categorical variable
sns.countplot(x ='sex', hue = "smoker", data = tips)
plt.show()
```

```python
# count plot along y axis
sns.countplot(y ='sex', hue = "smoker", data = tips)

plt.show()
```
