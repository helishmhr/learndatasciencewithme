# **Distribution Plot**

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
penguins=sns.load_dataset('penguins')
penguins
```

```python
tips = sns.load_dataset("tips")
tips
```

## **Histogram**
Plotted using distplot() or histplot()

```python
#basic histogram plot
sns.histplot(data = penguins, x = "flipper_length_mm")
plt.show()
```

```python
#choosing bin size
sns.histplot(data = penguins, x = "flipper_length_mm", binwidth=4)
plt.show()
```

```python
#fixing the no. of bins
sns.histplot(data = penguins, x = "flipper_length_mm", bins=20)
plt.show()
```

```python
#histogram for categorical variable
sns.histplot(data=tips, x="day")
plt.plot()
```

```python
#Conditioning on other variables
sns.histplot(penguins, x="flipper_length_mm", hue="species")
plt.show()
```

```python
#drawing individual distribution in a separate subplot by assigning the second variable to col or row
sns.displot(penguins, x="flipper_length_mm", col="sex")
plt.show()
```

## **Kernel density estimation**

```python
#generating approximate the underlying probability density function
sns.displot(penguins, x="flipper_length_mm", kind="kde")
plt.show()
```

```python
#choosing bandwidth to improve accuracy of representation
sns.displot(penguins, x="flipper_length_mm", kind="kde", bw_adjust=.25)
plt.show()
```

```python
#histogram with kde
sns.histplot(data = penguins, x = "flipper_length_mm", kde=True)
plt.show()
```

## **Pair Plot**

```python
#simple pair plot
sns.pairplot(tips)
plt.show()
```

```python
#pair plot with hue
sns.pairplot(tips, hue="day")
plt.show()
```

```python
#pairplot with specific variables
sns.pairplot(tips[['total_bill', 'tip', 'size']])
plt.show()
```
