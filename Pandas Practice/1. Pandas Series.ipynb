# **Pandas Series**

```python
import numpy as np
import pandas as pd
```

## **Creating Series**

```python
#creating series(without label) using lists
x=[10,20,30,40]
y= pd.Series(x)
print(y)
```

```python
#creating series(with label) using lists
y=pd.Series(x, index=['a','b','c','d'])
print(y)
```

```python
#Creating a Series(without label) from a NumPy Array
x = np.array(['g', 'e', 'e', 'k', 's'])

y = pd.Series(x)
print(y)
```

```python
#Creating a Series from a key-value objects(dictionary)
calories = {"day1": 420, "day2": 380, "day3": 390}
myvar = pd.Series(calories)
print(myvar)

#Note: The keys of the dictionary become the labels.
```

## **Accessing elements of Series**

```python
#accessing elements of series(without label)
print(y[0])
```

```python
#accessing elements of series(with label)
print(myvar['day1'])
```
