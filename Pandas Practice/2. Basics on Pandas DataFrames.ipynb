# **Basics on Pandas DataFrames**

```python
import numpy as np
import pandas as pd
```

## **Creating DataFrame**

```python
#creating DataFrame from a list
lst = ['John', 'Anna', 'Peter', 'Linda']
df = pd.DataFrame(lst)
print(df)
```

```python
#Creating a DataFrame from a List of Dictionaries  
data = [
    {'Name': 'John', 'Age': 28, 'City': 'New York', 'Salary': 65000 },
    {'Name': 'Anna', 'Age': 34, 'City': 'Paris', 'Salary': 70000  },
    {'Name': 'Peter', 'Age': 29, 'City': 'Berlin', 'Salary': 62000 },
]
df = pd.DataFrame(data)
print(df)
```

```python
#creating a data frame from dictionary
mydataset = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}

df = pd.DataFrame(mydataset)

print(df)
```

## **Accessing Specific Columns**

```python
#getting a specific column
print(df['Name'])
```

```python
#getting multiple columns
print(df[['Name', 'Age']])
```

## **Accessing Specific Rows**

```python
#getting a specific row of data frame
print(df.loc[0])
```

```python
#getting multiple rows
print(df.loc[[0, 1]])
```

## **Accessing Specific Rows and Columns**

```python
#getting rows and columns
print(df.loc[[0,1]][['Name','Age']])
```

## **Conditional Selections**

```python
#selecting persons with age>30
print(df[df['Age']>30])
```

```python
#selecting persons with age>30 and city=paris
print(df[(df['Age']>30)& (df['City']=='Paris')])
```

## **Named Indexes**

```python
#Giving own names to indices(naming each row with own name)
df=pd.DataFrame(mydataset, index=['p1', 'p2', 'p3','p4'])
print(df)
```

```python
#getting a specific row of data frame having named indices
print(df.loc['p1'])
```

## **Concatenating DataFrames**

```python
#Concatenating DataFrames Vertically
df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'C': [7, 8]})

res = pd.concat([df1, df2])
print(res)
```

```python
#Concatenating DataFrames Horizontally
res = pd.concat([df1, df2], axis=1)
print(res)
```

## **Merging DataFrames**

```python
# DataFrame 1: Employee information
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})

print(employees)
print()
print(salaries)
```

```python
#inner merge
res=pd.merge(employees,salaries, on='employee_id',how='inner')
print(res)
```

```python
#outer merge
res=pd.merge(employees,salaries, on='employee_id',how='outer')
print(res)
```

```python
#left merge
res=pd.merge(employees,salaries, on='employee_id',how='left')
print(res)
```

```python
#right merge
res=pd.merge(employees,salaries, on='employee_id',how='right')
print(res)
```

```python
#merging on different column names

# DataFrame 1: Employee information
employees = pd.DataFrame({
    'id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'emp_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})

res=pd.merge(employees,salaries,left_on='id',right_on='emp_id',how='inner')
print(res)
```

## **Apply Function**

```python
mydataset = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}

df = pd.DataFrame(mydataset)

print(df)
```

```python
#applying a custom or built-in function to each element of a Series or along an axis of a DataFrame
def multiplier(value):
    return value*2

df['Salary']=df['Salary'].apply(multiplier)
print(df)
```
