# **Data Cleaning**

```python
import numpy as np
import pandas as pd
```

## **Handling Missing values**

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, 5],
    'D': [1, 2, 3, 4, 5]
}
df = pd.DataFrame(data)
print(df)
```

```python
#checking missing values
print(df.isna())
```

```python
#Removing rows that contain at least one missing value
print(df.dropna())
```

```python
#filling missing values with 0
print(df.fillna(0))
```

```python
#Filling missing values Only For Specified Columns
print(df.fillna({'A':200,'C': 100}))
```

```python
#filling with mean
x = df["C"].mean()
df.fillna({"C": x})
```

**Note:**\
By default,all these method returns a new DataFrame, and will not change the original.\
If you want to change the original DataFrame, use the *inplace = True* argument

## **Changing Data type**

```python
data = {'Name': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'], 
        'Age': [25, 30, 22, 35, 28], 
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'], 
        'Salary': [50000, 55000, 40000, 70000, 48000],
       'Join Date':['2021-01-01', '2020-05-22', '2022-03-15', '2021-07-30', '2020-11-11'] }
df = pd.DataFrame(data)
print(df)
```

```python
#viewing data types of each column
print(df.dtypes)
```

```python
# Convert 'Age' column to float type
df['Age'] = df['Age'].astype(float)
print(df.dtypes)
```

```python
#Converting a Column to a DateTime Type
df['Join Date'] = pd.to_datetime(df['Join Date'])
print(df.dtypes)
```

## **Removing Duplicates**

```python
data = {
    "Name": ["Alice", "Bob", "Alice", "David"],
    "Age": [25, 30, 25, 40],
    "City": ["NY", "LA", "NY", "Chicago"]
}

df = pd.DataFrame(data)
print(df)
```

```python
#Discovering duplicates
print(df.duplicated())

#Returns True for every row that is a duplicate, otherwise False
```

```python
df.drop_duplicates(inplace = True)
print(df)
```
