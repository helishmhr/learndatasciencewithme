# **Customizating Plots**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Set Font Properties for Title and Labels
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

font1 = {'family':'serif','color':'blue','size':20}
font2 = {'family':'serif','color':'darkred','size':15}

plt.title("Simple Plot", fontdict = font1)
plt.xlabel("x-axis", fontdict = font2)
plt.ylabel("y-axix", fontdict = font2)

plt.plot(x, y)
plt.show()
```

```python
#adding grid into the plot
plt.plot(x, y)
plt.grid()
plt.show()
```

```python
#specifying which grid to display
plt.plot(x, y)
plt.grid(axis = 'x')
plt.show()

#Legal values are: 'x', 'y', and 'both'. Default value is 'both'.
```

```python
#customizing tick locations
plt.plot(x, y)
plt.xticks([0,2,4,6,8])
plt.yticks([0,10,20,30])
plt.show()
```

```python
#setting labels for ticks
plt.plot(x, y)
plt.xticks([0,2,4,6,8],['A','B','C','D','E'] )
plt.show()
```

```python
#multiple subplots in same figure
#create four different subplots in a 2x2 grid
#plot 1:
plt.subplot(2, 2, 1)
plt.plot(x,y)

#plot 2:
plt.subplot(2, 2, 2)
plt.plot(x,x)

#plot 3:
plt.subplot(2, 2, 3)
plt.plot(y,y)

#plot 4:
plt.subplot(2, 2, 4)
plt.plot(y,x)

plt.show()
```
