# **Line Chart**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#simple line chart
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y)

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Plot')

plt.show()
```

```python
#when x values are not provided its automatically generated as [1,2,...]
plt.plot(y)
plt.show()
```

```python
#Plotting with categorical variables
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
temperature = [22, 24, 23, 26, 25]

plt.plot(days, temperature, marker='o')
plt.title('Weekly Temperature')
plt.xlabel('Days')
plt.ylabel('Temperature (°C)')
plt.show()
```

```python
#multiple plots on same axes
plt.plot(x, y)
plt.plot(x, x)

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Plot')

plt.show()
```

```python
#plotting without a line
plt.plot(x, y, 'o')
plt.show()
```

```python
#changing the style of line
plt.plot(x, y, linestyle = 'dotted')
plt.show()
```

```python
plt.plot(x, y, linestyle = 'dashed')
plt.show()
```

```python
#changing color of line
plt.plot(x, y, color="#A81F03")
plt.show()
```

```python
#changing width of line
plt.plot(x, y, linewidth = '20.5')
plt.show()
```
