# **Matplotlib Application Interfaces**

Matplotlib has two major application interfaces, or styles of using the library:

1. An explicit "Axes" interface that uses methods on a Figure or Axes object to create other Artists, and build a visualization step by step. This has also been called an "object-oriented" interface.

2. An implicit "pyplot" interface that keeps track of the last Figure and Axes created, and adds Artists to the object it thinks the user wants.

## **The explicit "Axes" interface**

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#creating a simple plot
fig = plt.figure()
ax = fig.subplots()
ax.plot([1, 2, 3, 4], [0, 0.5, 1, 0.2])

plt.show()
```

```python
#customizing the plot
fig = plt.figure()
ax = fig.subplots()
ax.plot([1, 2, 3, 4], [0, 0.5, 1, 0.2])

ax.set_title("Curve")
ax.set_xlabel('x_label')
ax.set_ylabel('y_label')

plt.show()
```

```python
#creating multiple subplots

# Sample Data
x = np.linspace(0, 2 * np.pi, 400)
y1 = np.sin(x**2)
y2 = np.cos(x**2)

# Create subplots
fig, axs = plt.subplots(2, figsize=(7,4))
fig.suptitle('Creating Subplots in explicit interface')

# Plotting data on individual subplots
axs[0].plot(x, y1)
axs[1].plot(x, y2)

# Adding titles to subplots
axs[0].set_title('Sin Plot')
axs[1].set_title('Cos Plot')

# Adjust the layout
plt.tight_layout()

# Displaying the plot
plt.show()
```

# **The implicit "pyplot" interface**

```python
#creating a simple plot
plt.figure()
plt.subplot(1,1,1)
plt.plot([1, 2, 3, 4], [0, 0.5, 1, 0.2])
plt.show()
```

```python
#creating multiple plots

x1 = np.linspace(0, 10, 100)
# create a plot figure
plt.figure()


# create the first of two panels and set current axis
plt.subplot(2, 1, 1)   # (rows, columns, panel number)
plt.plot(x1, np.sin(x1))


# create the second of two panels and set current axis
plt.subplot(2, 1, 2)   # (rows, columns, panel number)
plt.plot(x1, np.cos(x1));

plt.show()
```
