# **Numpy random**

```python
import numpy as np
from numpy import random
```

## **Generating random numbers**

```python
#random integer from 0-100
print(random.randint(100))
```

```python
#random float between 0-1
print(random.rand())
```

```python
#random 1D array containing 5 integers from 0-100
x= random.randint(100, size=(5))
print(x)
```

```python
#random 2D array with 3 rows each row containing 5 integers from 0-100
x = random.randint(100, size=(3, 5))
print(x)
```

```python
#random 1D array containing 5 floats from 0-1
print(random.rand(5))
```

```python
#random 2D array with 3 rows each row containing 5 floats from 0-1
print(random.rand(3,5))
```

```python
#generating random number from array
x= random.choice([1,2,3,4,5])
print(x)
```

```python
#Generating a 1-D array that consists of the values in the array parameter (3, 5, 7, and 9):
x = random.choice([3, 5, 7, 9], size=(3))
print(x)
```

```python
#Generating a 2-D array that consists of the values in the array parameter (3, 5, 7, and 9):
x = random.choice([3, 5, 7, 9], size=(3,5))
print(x)
```

## **Random Distribution**

```python
#generate random numbers based on defined probabilities
x = random.choice([3, 5, 7, 9], p=[0.1, 0.3, 0.6, 0.0], size=(10))
print(x)
```

## **Random Permutations**

```python
#changing arrangement of elements in-place. i.e. in the array itself
arr= np.array([1,2,3])
random.shuffle(arr)
print(arr)
```

```python
#Generating a random permutation of elements of following array
arr= np.array([1,2,3])
new = random.permutation(arr)
print("Original array:", arr)
print("New array:",new)
```

## **Normal Distribution**

```python
#generating a random number from a standard normal distribution with mean = 0 and standard deviation = 1
x = random.normal(loc=0, scale=1, size=None)
print(x)
```

```python
#generating a 1-D array of five random numbers drawn from a normal distribution with mean = 0 and standard deviation = 1
arr = random.normal(size=5)
print(arr)
```

```python
#generating a 2-D array of size 2x3 of random numbers drawn from a normal distribution with mean = 1 and standard deviation = 2
x = random.normal(loc=1, scale=2, size=(2, 3))
print(x)
```

## **Binomial Distribution**

```python
#generating a 1-D array of 10 random numbers drawn from a binomial distribution with no. of trials(n)=10 and p=0.5
x= random.binomial(n=10, p=0.5, size=10)
print(x)
```

## **Multinomial Distribution**
(Generalization of Binomial distribution)

```python
x = random.multinomial(n=10, pvals=[1/6, 1/6, 1/6, 1/6, 1/6,1/6], size=3)
print(x)
```

## **Poisson Distribution**

```python
#generating a 1-D array of 10 random numbers drawn from a Poisson distribution with average rate=2
x = random.poisson(lam=2, size=10)
print(x)
```

## **Uniform Distribution**

```python
#generating 1D array of five random numbers in the range 10(inclusive) to 20(exclusive)
x = np.random.uniform(low=10, high=20, size=5)
print(x)
```

## **Logistic Distribution**

```python
x = random.logistic(loc=1, scale=2, size=(2, 3))
print(x)
```

## **Exponential Distribution**

```python
x = random.exponential(scale=2, size=(2, 3))
print(x)
```

## **Chi Squared Distribution**

```python
#generating a 2-D array of size 2x3 of random numbers drawn from a Chi squared distribution with degree of freedom=2
x = random.chisquare(df=2, size=(2, 3))
print(x)
```
